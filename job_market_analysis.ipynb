```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    r"C:\Users\sachingowda\Documents\job_market_analysis.csv",
    encoding="latin1"
)
```

```python
df.head()
```

```python
df['State'] = df['Location'].str.split(',').str[-1].str.strip()

state_jobs = df['State'].value_counts()

print(state_jobs.head(10))
```

```python
df.columns.tolist()
```

```python
for col in df.columns:
    print(col)
```

```python
[col for col in df.columns if 'salary' in col.lower()]
```

```python
[col for col in df.columns if 'salary' in col.lower()]
```

```python
state_min_salary = df.groupby('State')['Lower_Salary'].mean()

print(state_min_salary.round(2))
```

```python
state_max_salary = df.groupby('State')['Upper_Salary'].mean()

print(state_max_salary.round(2))
```

```python
salary_by_state = df.groupby('State').agg({
    'Lower_Salary': 'mean',
    'Upper_Salary': 'mean'
})

salary_by_state = salary_by_state.round(2)

display(salary_by_state)
```

```python
salary_by_state.plot(kind='bar', figsize=(14, 6))

plt.title('Average Minimum and Maximum Salary by State')
plt.xlabel('State')
plt.ylabel('Salary (USD thousands)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
salary_by_state.to_csv('salary_by_state.csv')
```

```python
average_salary = df.groupby('State')['Avg_SalaryK'].mean()

print(average_salary.round(2))
```

```python
top_salary_states = average_salary.sort_values(ascending=False).head(10)

print(top_salary_states)
```

```python
plt.figure(figsize=(12, 6))

top_salary_states.plot(kind='bar')

plt.title('Top 10 States by Average Salary')
plt.xlabel('State')
plt.ylabel('Average Salary (USD thousands)')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
top_salary_states.to_csv('top_salary_states.csv')
```

```python
industry_jobs = df['Industry'].value_counts().head(5)

print(industry_jobs)
```

```python
industry_jobs.plot(kind='bar', figsize=(10, 5))

plt.title('Top 5 Industries by Job Postings')
plt.xlabel('Industry')
plt.ylabel('Number of Job Postings')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
job_titles = df['Job Title'].value_counts()

print(job_titles.head(10))
```

```python
company_jobs = df['company_txt'].value_counts()

print(company_jobs.head(10))
```

```python
top_companies = company_jobs.head(10)

plt.figure(figsize=(12, 6))
top_companies.plot(kind='bar')

plt.title('Top 10 Companies with Most Job Postings')
plt.xlabel('Company Name')
plt.ylabel('Number of Job Postings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_companies.to_csv('top_10_companies.csv')
```

```python
job_titles = df['Job_Title'].value_counts()

print(job_titles.head(10))
```

```python
top_job_titles = job_titles.head(10)

top_job_titles.plot(kind='bar', figsize=(12, 6))

plt.title('Top 10 Job Titles by Job Postings')
plt.xlabel('Job Title')
plt.ylabel('Number of Job Postings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
job_salary = df.groupby('Job_Title')['Avg_SalaryK'].mean()

print(job_salary.sort_values(ascending=False).head(10).round(2))
```

```python
top_job_salary = job_salary.sort_values(ascending=False).head(10)

top_job_salary.plot(kind='bar', figsize=(12, 6))

plt.title('Top 10 Job Titles by Average Salary')
plt.xlabel('Job Title')
plt.ylabel('Average Salary (USD thousands)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_job_salary.to_csv('top_job_salary.csv')
```

```python
skills = ['Python', 'spark', 'aws', 'excel', 'sql_', 'tableau']

skill_counts = df[skills].sum().sort_values(ascending=False)

print(skill_counts)
```

```python
skill_counts.plot(kind='bar', figsize=(8, 5))

plt.title('Most Demanded Skills in Job Market')
plt.xlabel('Skills')
plt.ylabel('Number of Job Postings')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
education_salary = df.groupby('Degree')['Avg_SalaryK'].mean()

print(education_salary.round(2))
```

```python
education_salary.plot(kind='bar', figsize=(8, 5))

plt.title('Average Salary by Education Requirement')
plt.xlabel('Degree')
plt.ylabel('Average Salary (USD thousands)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
